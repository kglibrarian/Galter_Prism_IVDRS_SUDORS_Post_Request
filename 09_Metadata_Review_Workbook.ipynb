# Metadata review workbook (for Gretchen)

Read-only. Joins every published Prism record in the community to its row in the deposit workbook and writes
`output/prism/IVDRS_SUDORS_Metadata_Review.xlsx`:

- **Records**: one row per record with the Prism page, the DOI to cite, the file name(s), a link to the file on
  SharePoint, the original source, notes recorded while preparing the record, and yellow columns for the reviewer.
- **Files**: one row per file (the two county-report records have 16 each).

**SharePoint link:** set `SHAREPOINT_FILES_URL` to the web address of the SharePoint folder that holds the same
`ivdrs_reports/`, `sudors_reports/` and `articles/` folders as `data/files/`, written as a plain path, e.g.
`https://northwestern.sharepoint.com/sites/YOURSITE/Shared Documents/IVDRS/files`
(not a "Copy link" sharing link). Leave it as `None` to skip the links; the file names and folders are still listed.

```python
import importlib
import prism_deposit as pd_
pd_ = importlib.reload(pd_)
from paths import WORKBOOK, CONFIG_FILE, PRISM_OUT

ENV        = "production"
PRISM_URL  = "https://prism.northwestern.edu"
TOKEN_VAR  = "PRISM_PRODUCTION_TOKEN"
COMMUNITY_SLUG = "ivdrs-sudors"
SHAREPOINT_FILES_URL = None      # e.g. "https://northwestern.sharepoint.com/sites/YOURSITE/Shared Documents/IVDRS/files"
OUT_XLSX = PRISM_OUT / "IVDRS_SUDORS_Metadata_Review.xlsx"

env = pd_.load_config(CONFIG_FILE)
api = pd_.Prism(PRISM_URL, env.get(TOKEN_VAR, ""), verify=True)
reg = pd_.Registry(PRISM_OUT / f"{ENV}_registry.json")
community = api.get(f"/api/communities/{COMMUNITY_SLUG}")
assert community, f"No community '{COMMUNITY_SLUG}' on {PRISM_URL}"
records = pd_.community_records(api, community["id"])
print(pd_.save_review_workbook(records, reg, WORKBOOK, OUT_XLSX, SHAREPOINT_FILES_URL))
print("Saved", OUT_XLSX)
```
